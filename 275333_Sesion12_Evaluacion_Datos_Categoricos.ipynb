## Actividad 4 — Tipos de dato (30 pts)

```python
df['TotalCharges'].dtype
```

**4.1 — Investiga (responde en Markdown):**

`TotalCharges` contiene valores numéricos (montos en dólares), pero pandas la cargó como `object`, no como `float`. Investiga por qué — revisa si hay algún valor que no se vea como un número normal.

**_Tu respuesta:_** Existen 11 registros en los que TotalCharges contiene un valor vacío (espacios en blanco) en lugar de un valor numérico. Debido a que la columna contiene valores numéricos y estos valores de texto, pandas la interpreta como tipo object en lugar de float.

```python
df['TotalCharges'].value_counts()
```

```python
# 4.2 — Corrige el tipo de TotalCharges.
# Pista: pd.to_numeric() con el parámetro errors= te puede ayudar a identificar
# o manejar los valores problemáticos que encontraste en 4.1.
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
#comprobación de tipo de dato
df['TotalCharges'].dtype

```

```python
#Comprobación de valores
df['TotalCharges'].value_counts()
```

```python
#guardar el uso de memoria antes de la transformación para su posterior análisis
memoria_antes = df.memory_usage(deep=True)
```

```python
# 4.3 — Convierte a category las columnas categóricas que, según lo que calculaste
# en la Actividad 3, tengan cardinalidad baja y valores fijos.
# Verifica con .dtypes que el cambio se aplicó correctamente.

# Detectar columnas tipo object
cat_cols = df.select_dtypes(include=['object']).columns
#Convertir columnas object a category, excepto customerID
for col in cat_cols:
    if col != 'customerID':
        df[col] = df[col].astype('category')
```

```python
#Comprobación de los tipos de datos corregidos
df.dtypes
```

```python
#comparar el uso de memoria antes y despues del cambio de tipo de dato

memoria_despues = df.memory_usage(deep=True)

comparacion = pd.DataFrame({
    'Antes': memoria_antes,
    'Después': memoria_despues
})

#agregar fila de totales
comparacion.loc['Total'] = comparacion.sum()

#calcular ahorro
comparacion['Ahorro'] = (
    comparacion['Antes'] - comparacion['Después']
)

#agregar fila de ahorro %
comparacion['Ahorro %'] = (
    (comparacion['Antes'] - comparacion['Después']) / comparacion['Antes'] * 100
).round(2)

comparacion
```

**Conclusión:** La conversión de TotalCharges a tipo numérico permitió corregir su formato y tratar los valores problemáticos como valores faltantes. Posteriormente, la conversión de las variables categóricas de baja cardinalidad a tipo category produjo una reducción significativa en el uso de memoria. El DataFrame pasó de utilizar 6,656,929 bytes a 757,481 bytes, lo que representa un ahorro de 5,899,448 bytes, equivalente al 88.62% del uso de memoria original. Esto demuestra que el uso de category es una estrategia eficiente para optimizar el almacenamiento de variables categóricas con pocos valores únicos.
