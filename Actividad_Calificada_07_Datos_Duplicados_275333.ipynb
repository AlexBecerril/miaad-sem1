## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# En este apartado se calcula el número de duplicados exactos
# del DataFrame "hoteles", considerando todas sus columnas

# Marca como duplicados todas las apariciones posteriores a la primera
print("Registros duplicados (keep=first):", hoteles.duplicated(keep='first').sum())

# Marca como duplicados todas las apariciones anteriores a la última
print("Registros duplicados (keep=last):", hoteles.duplicated(keep='last').sum())

# Marca como duplicadas todas las apariciones de un registro que tiene duplicados
print("Totalidad de registros duplicados (keep=False):", hoteles.duplicated(keep=False).sum())
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# En este apartado se calcula el total de registros sobrantes,
# qué porcentaje representan del total de registros y
# cuántos grupos de registros repetidos existen

# Se cuentan todos los registros que pertenecen a un grupo duplicado,
# incluyendo la primera aparición de cada grupo
total_duplicados = hoteles.duplicated(keep=False).sum()

# Se cuentan únicamente las apariciones posteriores a la primera,
# es decir, los registros que serían sobrantes si se conservara
# la primera aparición de cada grupo
total_sobrantes = hoteles.duplicated(keep='first').sum()

# Se obtiene el número total de registros del DataFrame
total_registros = len(hoteles)

# Se calcula qué proporción del total de registros corresponde a las filas sobrantes
porcentaje_sobrantes = total_sobrantes/total_registros

# Se convierte la proporción a porcentaje y se redondea a dos decimales
print("El porcentaje de filas sobrantes respecto al total es:", round(porcentaje_sobrantes * 100, 2), "%")

# La diferencia entre todos los registros pertenecientes a grupos
# duplicados y las filas sobrantes representa el número de grupos repetidos
total_repetidos = total_duplicados - total_sobrantes
print("El número de grupos repetidos es:", total_repetidos)
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# En este apartado se crea una copia del DataFrame original y se agrega
# una columna booleana que identifica las filas sobrantes
hoteles_copia = hoteles.copy()

# Se identifican como duplicados las apariciones posteriores a la primera
hoteles_copia['es_duplicado'] = hoteles_copia.duplicated(keep='first')

# Se agrupan los registros por hotel y se suman los valores True
# para obtener el número de filas sobrantes de cada hotel
duplicados_por_hotel = hoteles_copia.groupby('hotel')['es_duplicado'].sum()
duplicados_por_hotel
```

```python
# Se calcula el total de reservas de cada hotel
registros_por_hotel = hoteles_copia.groupby('hotel').size()
registros_por_hotel
```

```python
# Se calcula el porcentaje de filas sobrantes respecto al total de reservas de cada hotel
porcentaje_por_hotel = round(duplicados_por_hotel / registros_por_hotel * 100, 2)
porcentaje_por_hotel
```

Filtra las filas involucradas en duplicados (`keep=False`), **ordénalas** con `sort_values()` para que las copias queden juntas y muestra las primeras 12 filas con estas columnas:

### 2.2 Tamaño de los grupos (6 pts)

Usa `value_counts()` sobre el `DataFrame` completo para contar cuántas veces aparece cada combinación de filas. Considera que las columnas `agent` y `company` tienen faltantes (revisa el parámetro `dropna`).

Con ese resultado:

1. Muestra **cuántos grupos** hay de cada tamaño (2 copias, 3 copias, …). Basta con los 10 tamaños más frecuentes.

```python
# Se cuenta cuántas veces aparece cada combinación exacta de valores,
# incluyendo los valores NaN de las columnas agent y company.
conteo_grupos = hoteles.value_counts(dropna=False)

# Se eliminan las combinaciones que aparecen una sola vez,
# ya que no forman parte de un grupo repetido.
conteo_grupos_repetidos = conteo_grupos[conteo_grupos > 1]

# Se cuenta cuántos grupos existen para cada tamaño.
grupos_por_tamano = conteo_grupos_repetidos.value_counts()

# Se muestran los 10 tamaños de grupo más frecuentes.
grupos_por_tamano.head(10)
```

```python
# Se obtiene el número máximo de apariciones de una misma combinación.
tamano_grupo_maximo = conteo_grupos.max()

print("El tamaño del grupo más grande es:", tamano_grupo_maximo)

# Se obtiene la combinación de valores correspondiente al grupo más grande.
grupo_mas_grande = conteo_grupos[
    conteo_grupos == tamano_grupo_maximo
]

grupo_mas_grande
```

```python
# Se calcula la proporción de cada categoría de is_canceled
# dentro de las filas involucradas en duplicados
is_canceled_prop_duplicados = round(hoteles_duplicados['is_canceled'].value_counts(dropna=False, normalize=True) * 100, 2).rename('proporcion_duplicados')

# Se calcula la proporción de cada categoría de is_canceled
# dentro de todo del dataset
is_canceled_prop_total = round(hoteles['is_canceled'].value_counts(dropna=False, normalize=True) * 100, 2).rename('proporcion_total')

# Se concatenan ambas distribuciones para compararlas lado a lado
is_canceled_prop_comparacion = pd.concat([is_canceled_prop_duplicados, is_canceled_prop_total], axis=1).fillna(0)

is_canceled_prop_comparacion
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Métricas por hotel para el dataset original
resumen_original = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean()

# Número de reservas por hotel
resumen_original['reservas'] = hoteles.groupby('hotel').size()

# Se convierte hotel nuevamente en una columna
resumen_original = resumen_original.reset_index()

resumen_original
```

```python
# Métricas por hotel para el dataset filtrado
resumen_limpio = hoteles_limpio.groupby('hotel')[['is_canceled', 'adr']].mean()

# Número de reservas por hotel
resumen_limpio['reservas'] = hoteles_limpio.groupby('hotel').size()

# Se convierte hotel nuevamente en una columna
resumen_limpio = resumen_limpio.reset_index()

resumen_limpio
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Se agrega una columna para identificar si cada resumen corresponde
# al dataset original o al dataset sin duplicados
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

# Se combinan ambos DataFrames en uno solo
comparacion = pd.concat([resumen_original, resumen_limpio])

# Se ordenan los resultados por hotel y se renumera el índice
comparacion = comparacion.sort_values(by='hotel').reset_index(drop=True)

comparacion
```
