### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.

```python
# Se calcula la duración de cada viaje como la diferencia
# entre la fecha y hora de dropoff y pickup
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

# Se convierte la duración a minutos utilizando el total de segundos
# y dividiendo entre 60
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60

# Se obtiene el resumen estadístico de la duración de los viajes en minutos
taxis['duracion_min'].describe()
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Se extrae la hora del pickup
taxis['hora'] = taxis['pickup'].dt.hour

# se extrae el día de la semana como número
taxis['dia_num'] = taxis['pickup'].dt.dayofweek

# Se extrae el nombre del día
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()

# True si el viaje inicia en sábado (5) o domingo (6)
taxis['fin_semana'] = taxis['dia_num'].isin([5, 6])

# Se extrae la fecha del pickup sin la hora
taxis['fecha'] = taxis['pickup'].dt.date

# Se compara la fecha de pickup con la fecha de dropoff
# Es True cuando el viaje termina en una fecha diferente
taxis['cruza_medianoche'] = (
    taxis['pickup'].dt.date != taxis['dropoff'].dt.date
)

# Se cuentan cuántos viajes cruzan la medianoche
print(
    'Viajes que cruzan la medianoche:',
    taxis['cruza_medianoche'].sum()
)
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# Se agrupan los viajes por hora y se calcula el número de viajes
# y la mediana de la duración de los viajes
df_filtrado = pd.concat([
    taxis.groupby('hora')['duracion_min'].count().rename('cantidad'),
    taxis.groupby('hora')['duracion_min'].median().rename('mediana_duracion')
], axis=1).reset_index(drop=True)

df_filtrado
```

```python
# Se identifica la hora con mayor cantidad de viajes
print("Hora con más viajes:")
print( df_filtrado.loc[df_filtrado['cantidad'].idxmax()])

# Se identifica la hora con menor cantidad de viajes
print("\nHora con menos viajes:")
print(df_filtrado.loc[df_filtrado['cantidad'].idxmin()])
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Se agrupan los viajes por número y nombre del día de la semana
# y se calcula el número de viajes, la mediana de la duración,
# la mediana de la tarifa y el número de fechas distintas
df_filtrado = pd.concat([
    taxis.groupby(['dia_num', 'dia_nombre'])['duracion_min'].count().rename('cantidad_viajes'),

    taxis.groupby(['dia_num', 'dia_nombre'])['duracion_min'].median().rename('mediana_duracion'),

    taxis.groupby(['dia_num', 'dia_nombre'])['fare'].median().rename('mediana_tarifa'),

    taxis.groupby(['dia_num', 'dia_nombre'])['fecha'].nunique().rename('numero_fechas')
], axis=1)

# Se calcula el promedio de viajes por fecha para cada día de la semana,
# dividiendo el número total de viajes entre el número de fechas distintas
df_filtrado['promedio_viajes_por_fecha'] = (df_filtrado['cantidad_viajes'] / df_filtrado['numero_fechas'])

# Se muestra el DataFrame con las métricas calculadas para cada día
df_filtrado
```
