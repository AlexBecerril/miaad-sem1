## Actividad 9 — Práctica integradora: checklist de profiling

*Dataset: muestra real de Customer Personality Analysis, con 2 elementos inyectados y marcados a propósito (una fila duplicada y un valor de tipo incorrecto) para poder practicar el checklist completo con un resultado garantizado.*

Aplica el checklist completo, en orden, sobre `df_practica`:

1. Revisa `.dtypes` e identifica qué columna tiene un problema de tipo, corrígela con `pd.to_numeric(errors='coerce')`
2. Cuenta las filas duplicadas y elimínalas con `.drop_duplicates()`
3. Cuenta los valores faltantes por columna con `.isnull().sum()` (incluyendo el que se generó en el paso 1)
4. Revisa `.unique()` sobre `Marital_Status` y decide si necesita normalización

Al final, escribe un breve "reporte de profiling" (3-4 líneas) resumiendo qué encontraste y qué decidiste.

```python
# Muestra real con 2 elementos inyectados (marcados abajo)
df_practica = df_marketing.sample(15, random_state=3).reset_index(drop=True).copy()

# Elemento inyectado 1: una fila duplicada
df_practica = pd.concat([df_practica, df_practica.iloc[[2]]], ignore_index=True)

# Elemento inyectado 2: un valor de tipo incorrecto en Income
df_practica['Income'] = df_practica['Income'].astype(object)
df_practica.loc[5, 'Income'] = 'sesenta mil'

df_practica[['ID', 'Marital_Status', 'Income']]
```

```python
# Paso 0 - Estandarización de nombres de columnas
#convertir todas las columnas a snake case y minusculas
df_practica.columns = df_practica.columns.str.replace(r'(?<!^)(?=[A-Z])', '_', regex=True).str.replace(r'_+', '_', regex=True).str.lower()
#Reemplazar nombres de columnas especificas para seguir la convensión de la linea anterior
df_practica = df_practica.rename(columns={'i_d': 'id', 'kidhome': 'kid_home', 'teenhome': 'teen_home'})
df_practica.columns
```

```python
# Paso 1 — ajuste de tipos
print(df_practica.dtypes)

print('\nExisten dos columnas con el tipo de dato incorrecto: income y dt_customer')

#Conversión a formato fecha de la columna "date_added"
df_practica['dt_customer'] = pd.to_datetime(df_practica['dt_customer'], format='mixed')
df_practica['income'] = pd.to_numeric(df_practica['income'], errors='coerce')

print('\nTipos de datos despues de la conversión:')
print(df_practica.dtypes)
```

```python
# Paso 2 — duplicados
print('número de filas originales: ', len(df_practica))
print('número de filas duplicadas: ', df_practica.duplicated().sum())
df_practica = df_practica.drop_duplicates()
print('número de filas sin duplicados: ', len(df_practica))
```

```python
# Paso 3 — valores faltantes
print('Valores faltantes por columna:\n', df_practica.isnull().sum())
df_practica_nulos = df_practica.isnull().any(axis=1).sum()
print('\nFilas con al menos un valor faltante:', df_practica_nulos)
```

```python
# Paso 4 — exploración categórica
print('Valores unicos iniciales de "marital_status":')
print(df_practica['marital_status'].value_counts())
print('\nLa columna "marital_status" no requiere normalización')
```

**Tu reporte de profiling:**
1. Dentro del conjunto de datos "df_practica" existían dos columnas con un tipo de dato incorrecto: "income", de tipo Object, fue convertida a tipo numeric, y "dt_customer" fue convertida a tipo datetime.
2. Dentro del conjunto de datos existían 15 filas únicas y una duplicada, la cual fue removida.
3. Al hacer la conversión de "income" a tipo numérico, se generó un valor faltante dentro de una de las filas, siendo este el único valor faltante dentro de todo el conjunto de datos.
4. Los valores únicos de "marital_status" corresponden todos a valores reales. No se detectó ningún valor atípico; por lo tanto, se determinó que no se requiere ningún tipo de normalización.
